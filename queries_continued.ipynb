```python
from dotenv import load_dotenv
load_dotenv('.env')
```

```python
from langchain_mcp_adapters.client import MultiServerMCPClient
from langgraph.prebuilt import create_react_agent

client = MultiServerMCPClient(
    {
        "database": {
            "url": "http://localhost:8000/mcp",
            "transport": "streamable_http",
        }
    }
)
tools = await client.get_tools()
agent = create_react_agent(
    "openai:gpt-5",
    tools
)
```

```python
tools
```

```python
vehicle_query = await agent.ainvoke(
    {"messages": [{"role": "user", "content": "How many electric vehicles were purchased in 2016 in Washington State?"}]}
)
```

```python
vehicle_query
```

```python
followup_query = await agent.ainvoke(
    {"messages": [{"role": "user", "content": "How did the range of electric vehicles sold change over model year in Washington State?"}]}
)
```
